```python
def primes_under(j):
    my_primes_under = [2,3]
    n = 1
    while 6*n - 1 < j:
        if is_prime_aux(6*n - 1):
            my_primes_under.append(6*n - 1)
        if is_prime_aux(6*n + 1) and 6*n + 1 <j:
            my_primes_under.append(6*n + 1)
        n = n + 1
    return my_primes_under
```
