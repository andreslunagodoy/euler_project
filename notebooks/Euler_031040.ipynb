**Problem 31: Coin Sums**

In the United Kingdom the currency is made up of pound (£) and pence (p). There are eight coins in general circulation:

> 1p, 2p, 5p, 10p, 20p, 50p, £1 (100p), and £2 (200p).

It is possible to make £2 in the following way:

> 1×£1 + 1×50p + 2×20p + 1×5p + 1×2p + 3×1p

How many different ways can £2 be made using any number of coins?

```python
# This is Euler project's problem 31
coin_values = [1,2,5,10,20,50,100,200]

def partitions(n,max_val):
    part = 0
    if n < 0:
        return 0
    elif n == 0:
        return 1
    else:
        for val in coin_values:
            if val > max_val:
                break
            else:
                part += partitions(n - val,val)
    return part
partitions(200,200)
```

**Problem 32: Pandigital Products**

We shall say that an $n$-digit number is pandigital if it makes use of all the digits $1$ to $n$ exactly once; for example, the $5$-digit number, $15234$, is $1$ through $5$ pandigital.

The product $7254$ is unusual, as the identity, $39 \times 186 = 7254$, containing multiplicand, multiplier, and product is $1$ through $9$ pandigital.

Find the sum of all products whose multiplicand/multiplier/product identity can be written as a $1$ through $9$ pandigital.

HINT: Some products can be obtained in more than one way so be sure to only include it once in your sum.

```python
# This is Euler project's problem 32
all_digits = [1,2,3,4,5,6,7,8,9]
triples = []
for perm in permutations(all_digits,2):
    obj = all_digits.copy()
    for x in perm:
        obj.remove(x)
    for perm2 in permutations(obj,3):
        obj2 = obj.copy()
        for x in perm2:
            obj2.remove(x)
        for perm3 in permutations(obj2,4):
            triples.append([digits_to_integer(perm),digits_to_integer(perm2),digits_to_integer(perm3)])
        
for perm in permutations(all_digits,1):
    obj = all_digits.copy()
    for x in perm:
        obj.remove(x)
    for perm2 in permutations(obj,4):
        obj2 = obj.copy()
        for x in perm2:
            obj2.remove(x)
        for perm3 in permutations(obj2,4):
            triples.append([digits_to_integer(perm),digits_to_integer(perm2),digits_to_integer(perm3)])

candidates = []
for triple in triples:
    if triple[0]*triple[1] == triple[2]:
        candidates.append(triple[2])
candidates=list(set(candidates))
total_sum_list(candidates)
```

**Problem 33: Digit Cancelling Fractions**

The fraction $49/98$ is a curious fraction, as an inexperienced mathematician in attempting to simplify it may incorrectly believe that $49/98 = 4/8$, which is correct, is obtained by cancelling the $9$s.

We shall consider fractions like, $30/50 = 3/5$, to be trivial examples.

There are exactly four non-trivial examples of this type of fraction, less than one in value, and containing two digits in the numerator and denominator.

If the product of these four fractions is given in its lowest common terms, find the value of the denominator.

```python
# This is Euler project's problem 33
all_fractions = []
for i in range(11,100):
    for j in range(11,i):
        all_fractions.append([j,i])
        
def is_frac_reducible(frac):
    num_set = set(to_digits(frac[0]))
    den_set = set(to_digits(frac[1]))
    if len(num_set.intersection(den_set)) > 0  and list(num_set.intersection(den_set)).pop() != 0:
        return True
    else:
        return False
    
all_reducible_fractions = []
for frac in all_fractions:
    if is_frac_reducible(frac):
        all_reducible_fractions.append(frac)

def reduce_frac(frac):
    num_digits = to_digits(frac[0])
    den_digits = to_digits(frac[1])
    inter = list(set(num_digits).intersection(set(den_digits))).pop()
    num_digits.remove(inter)
    den_digits.remove(inter)
    return [num_digits[0],den_digits[0]]

candidates = []
for frac in all_reducible_fractions:
    reduced = reduce_frac(frac)
    if frac[0]*reduced[1]==frac[1]*reduced[0]:
        candidates.append(frac)
my_prod_num = 1
my_prod_den = 1
for frac in candidates:
    my_prod_num *= frac[0]
    my_prod_den *= frac[1]
my_prod_den / my_prod_num
```

**Problem 34: Digit Factorials**

$145$ is a curious number, as $1! + 4! + 5! = 1 + 24 + 120 = 145$.

Find the sum of all numbers which are equal to the sum of the factorial of their digits.

Note: As $1! = 1$ and $2! = 2$ are not sums they are not included.

```python
# This is Euler project's problem 34
my_num=[]
for i in range(10,2600000):
    aux=to_digits(i)
    aux2=0
    for j in aux:
        aux2 += factorial(j)
    if aux2 == i:
        my_num.append(i)
total_sum_list(my_num)
```

**Problem 37: Truncatable Primes**

The number $3797$ has an interesting property. Being prime itself, it is possible to continuously remove digits from left to right, and remain prime at each stage: $3797$, $797$, $97$, and $7$. Similarly we can work from right to left: $3797$, $379$, $37$, and $3$.

Find the sum of the only eleven primes that are both truncatable from left to right and right to left.

NOTE: $2$, $3$, $5$, and $7$ are not considered to be truncatable primes.

```python
# This is Euler project's problem 37
def is_truncatable_left(n):
    digits = to_digits(n)
    while len(digits) > 0:
        if not is_prime_aux(digits_to_integer(digits)):
            return False
        else:
            digits.pop(0)
    return True

def is_truncatable_right(n):
    digits = to_digits(n)
    while len(digits) > 0:
        if not is_prime_aux(digits_to_integer(digits)):
            return False
        else:
            digits.pop()
    return True

trunc = []
index = 4
while len(trunc) < 11:
    n = my_list_of_primes[index]
    if is_truncatable_left(n) and is_truncatable_right(n):
        trunc.append(n)
    index += 1
total_sum_list(trunc)
```

**Problem 38: Pandigital Multiples**

Take the number $192$ and multiply it by each of $1$, $2$, and $3$:

$$\begin{align} 192 \times 1 &= 192\\ 192 \times 2 &= 384\\ 192 \times 3 &= 576 \end{align}$$

By concatenating each product we get the $1$ to $9$ pandigital, $192384576$. We will call $192384576$ the concatenated product of $192$ and $(1,2,3)$.

The same can be achieved by starting with $9$ and multiplying by $1$, $2$, $3$, $4$, and $5$, giving the pandigital, $918273645$, which is the concatenated product of $9$ and $(1,2,3,4,5)$.

What is the largest $1$ to $9$ pandigital $9$-digit number that can be formed as the concatenated product of an integer with $(1,2,\dots,n)$ where $n > 1$?

```python
# This is Euler project's problem 38
def is_pandigital(n):
    digits = to_digits(n)
    digits.sort()
    if digits == [1,2,3,4,5,6,7,8,9]:
        return True
    else:
        return False

max_pandigital = 0
my_seed = 0
for seed in range(1,10000):
    my_chain = []
    index = 1
    while len(my_chain) < 9:
        my_chain += to_digits(seed*index)
        index += 1
        my_number = digits_to_integer(my_chain)
    if (is_pandigital(my_number)) and my_number > max_pandigital:
        max_pandigital = my_number
        my_seed = seed
print([max_pandigital,my_seed])
```

**Problem 39: Integer Right Triangles**

If $p$ is the perimeter of a right angle triangle with integral length sides, $\{a, b, c\}$, there are exactly three solutions for $p = 120$:

$$\{20,48,52\}, \{24,45,51\}, \{30,40,50\}$$

For which value of $p \le 1000$, is the number of solutions maximised?

```python
# This is Euler project's problem 39
max_counter=0
max_num=0
for perimeter in range(12,1001):
    counter=0
    for hyp in range(perimeter//3+1,perimeter//2):
        for cat1 in range(hyp//2+1,hyp):
            cat2 = perimeter - hyp - cat1
            if cat1 > cat2:
                if hyp**2==cat1**2+cat2**2:
                    counter += 1
    if counter > max_counter:
        max_counter = counter
        max_num = perimeter
max_num
```
